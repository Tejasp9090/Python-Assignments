```python
#2. Write a Python program to calculate the element-wise sum of two tuples.
Tuple_1 = (1, 2, 3)
Tuple_2 = (4, 5, 6)
result=tuple(a + b for a,b in zip (Tuple_1, Tuple_2))
print(result)
```

```python
#3. Write a Python program to find all values that occur more than once in a tuple.
s_3=(1, 2, 2, 3, 4, 4, 4)
dup =()
for item in s_3:
  if s_3.count(item)>1 and item not in dup:
    dup+=(item,)
print("Repeated values = ",dup)
```

```python
#7. Write a Python program to calculate column-wise sums of a tuple-based matrix.
s_7=((1, 2, 3), (4, 5, 6), (7, 8, 9))
colums=zip(*s_7)

column_sums=[sum(column) for column in colums]
print("column_sums=",column_sums)
```
