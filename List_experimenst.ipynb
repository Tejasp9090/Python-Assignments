```python
#4. Write a Python program to count even and odd numbers in a list
l4=[1, 2, 3, 4, 5, 6]

even=0
odd=0

for num in l4:
  if num %2==0:
    even+=1
  else:
    odd+=1

print('Even:',even)
print('Odd:',odd)
```

```python
#5. Write a Python program to remove duplicate elements from a list.

l5=[1,2,2,3,1,4]
l_5=[]

for num in l5:
  if num not in l_5:
    l_5.append(num)

print(l_5)


```

```python
#12. Write a Python program to delete all occurrences of a given element from a list.
L12 = [1, 2, 3, 2, 4, 2]
delete = 2

result = []

for num in L12:
    if num != delete:
        result.append(num)

print(result)
```

```python
#13. Write a Python program to separate positive and negative numbers from a list.
s13=[-5, 3, -2, 7, 0, -1]
positive_index=[]
Negative_index=[]
for i in s13:
  if i>0:
      positive_index.append(i)
  elif (i<0):
      Negative_index.append(i)

print("positive_index", positive_index)
print("Negative_index", Negative_index)

```

```python
#14. Write a Python program to find common elements between two lists.
List_1 = [1, 2, 3, 4]
List_2 = [3, 4, 5, 6]

common=[]

for num in List_1:
  if num in List_2:
    common.append(num)

print('Common elements Are :',common)
```
